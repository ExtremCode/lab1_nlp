```python
symbs = "йцукенгшщзхъфывапролджэячсмитьбю"
symbs += symbs.upper()
symbs += ".,?!1234567890"
symbs = set(symbs)
```

```python
def filter_triple(tokens, lemmas, upos):
    nt, nl, nu = [], [], []
    for t, l, u in zip(tokens, lemmas, upos):
        if not symbs.isdisjoint(t):
            nt.append(t)
            nl.append(l)
            nu.append(u)
    return nt, nl, nu


res = [
    filter_triple(t, l, u)
    for t, l, u in zip(exdf["tokens"], exdf["lemmas"], exdf["upos"])
]
exdf["tokens"], exdf["lemmas"], exdf["upos"] = map(list, zip(*res))
```

```python
symbs = "йцукенгшщзхъфывапролджэячсмитьбю"
symbs += symbs.upper()
symbs += "1234567890"
symbs = set(symbs)
symbs.issuperset("ежик")
```

```python
def preprocessing(dataframe):
    df = dataframe.copy()
    df["tokens"] = df["tokens"].apply(
        lambda arr: [s.replace("ё", "е").replace("Ё", "Е") for s in arr]
    )
    df["lemmas"] = df["lemmas"].apply(
        lambda arr: [s.lower().replace("ё", "е").replace("Ё", "Е") for s in arr]
    )
    symbs = "йцукенгшщзхъфывапролджэячсмитьбю"
    symbs += symbs.upper()
    symbs += "1234567890"
    symbs = set(symbs)
    def filter_triple(tokens, lemmas, upos):
        nonlocal symbs
        nt, nl, nu = [], [], []
        for t, l, u in zip(tokens, lemmas, upos):
            if symbs.issuperset(t):
                nt.append(t)
                nl.append(l)
                nu.append(u)
        return nt, nl, nu


    res = [
        filter_triple(t, l, u)
        for t, l, u in zip(df["tokens"], df["lemmas"], df["upos"])
    ]
    df["tokens"], df["lemmas"], df["upos"] = map(list, zip(*res))
    return df
```
